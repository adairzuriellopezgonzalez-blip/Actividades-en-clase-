```python
# Ejercicio 3. Desigualdad triangular y formula de Heron
a = int(input("Ingresa el primer lado: "))
b = int(input("Ingresa el segundo lado: "))
c = int(input("Ingresa el tercer lado: "))
if a <= 0 or b <= 0 or c <= 0:
    print("Los lados deben ser positivos")
elif a + b > c and a + c > b and b + c > a:
    print("Los lados forman un triangulo")
    if a == b and b == c:
        print("El triangulo es equilatero")
    elif a == b or a == c or b == c:
        print("El triangulo es isosceles")
    else:
        print("El triangulo es escaleno")
    s = (a + b + c) / 2
    area = (s * (s - a) * (s - b) * (s - c)) ** 0.5
    print("El semiperimetro es:", s)
    print("El area del triangulo es:", area)
else:
    print("Los valores no forman un triangulo")
```

```python
# Ejercicio 6. Energia mecanica total
# Guardamos los datos del vehiculo
m = 1200
v = 25
g = 9.81
h = 15

# Calculamos la energia cinetica
Ec = (1/2) * m * v**2

# Calculamos la energia potencial
Ep = m * g * h

# Sumamos las energias
Em = Ec + Ep
# Mostramos los resultados
print("Energia cinetica:", Ec, "Joules")
print("Energia potencial:", Ep, "Joules")
print("Energia mecanica total:", Em, "Joules")
```
