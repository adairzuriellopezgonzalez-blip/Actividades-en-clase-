```python
# Formula de Heron de Alejandria
a = int(input("Ingresa el primer lado: "))
b = int(input("Ingresa el segundo lado: "))
c = int(input("Ingresa el tercer lado: "))
s = (a+b+c)/2
A = (s*(s-a)*(s-b)*(s-c))**0.5
print("El semiperimetro del triangulo es:", s)
print("El area del triangulo es:", A)
```

```python
import math

# Datos del elipsoide
a = 6378137.0
c = 6356752.314245
# Calcular la excentricidad
e2 = 1 - (c**2 / a**2)
e = math.sqrt(e2)
# Calcular la superficie del elipsoide
Sobl = 2 * math.pi * a**2 * (1 + ((1 - e2) / e) * math.atanh(e))
# Calcular la superficie de la esfera
r = 6371 * 1000
Sesfera = 4 * math.pi * r**2
# Comparar las superficies
diferencia = Sobl - Sesfera
porcentaje = (diferencia / Sesfera) * 100
# Mostrar resultados
print("Superficie del elipsoide:", Sobl, "m²")
print("Superficie de la esfera:", Sesfera, "m²")
print("Diferencia entre superficies:", diferencia, "m²")
print("Diferencia porcentual:", porcentaje, "%")
```
